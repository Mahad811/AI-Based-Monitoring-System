````python
# ============================================================================
# CELL 3.5: Robust Image Loader (CRITICAL - prevents all OpenCV errors)
# ============================================================================
import cv2
import numpy as np
from PIL import Image
import ultralytics.utils.patches as yolo_patches

print("🔧 Installing bulletproof image loader...")

_FAILED_IMAGES = set()

def robust_imread(filename, flags=cv2.IMREAD_COLOR):
    """
    Robust image loader that ALWAYS returns a valid numpy array.
    Falls back to PIL, then to dummy image if everything fails.
    """
    filepath = str(filename)
    
    # Method 1: Try OpenCV with explicit buffer
    try:
        with open(filepath, 'rb') as f:
            file_bytes = np.frombuffer(f.read(), dtype=np.uint8)
        img = cv2.imdecode(file_bytes, flags)
        if img is not None and isinstance(img, np.ndarray) and img.size > 0:
            # Ensure 3 channels
            if img.ndim == 2:
                img = cv2.cvtColor(img, cv2.COLOR_GRAY2BGR)
            elif img.ndim == 3 and img.shape[2] == 1:
                img = cv2.cvtColor(img, cv2.COLOR_GRAY2BGR)
            return img
    except Exception:
        pass
    
    # Method 2: Try PIL
    try:
        pil_img = Image.open(filepath).convert('RGB')
        img = np.array(pil_img)
        img = cv2.cvtColor(img, cv2.COLOR_RGB2BGR)  # PIL uses RGB, OpenCV uses BGR
        if img is not None and isinstance(img, np.ndarray) and img.size > 0:
            return img
    except Exception:
        pass
    
    # Method 3: Return black dummy image (training continues)
    if filepath not in _FAILED_IMAGES:
        print(f"⚠️  Using dummy image for: {filepath}")
        _FAILED_IMAGES.add(filepath)
    
    # Return valid 640x640 black image
    return np.zeros((640, 640, 3), dtype=np.uint8)

# Patch Ultralytics image readers
yolo_patches.imread = robust_imread

try:
    import ultralytics.data.base as yolo_base
    yolo_base.imread = robust_imread
except Exception:
    pass

print("✅ Robust image loader installed!")
print("   - Uses OpenCV with explicit buffer")
print("   - Falls back to PIL if OpenCV fails")
print("   - Never crashes - always returns valid array")
````

````python
# ============================================================================
# CELL 3.6: Helper utilities (latest run lookup)
# ============================================================================
from pathlib import Path

PROJECT_DIR = Path("seizure_detection")

def get_latest_run(project_dir: Path = PROJECT_DIR):
    """Return the most recent training run directory (or None)."""
    if not project_dir.exists():
        return None
    runs = sorted(
        [d for d in project_dir.iterdir() if d.is_dir()],
        key=lambda d: d.stat().st_mtime,
        reverse=True,
    )
    return runs[0] if runs else None

print("✅ Helper utilities loaded. Later cells will auto-detect the latest run.")
````

````python
# ============================================================================
# ULTIMATE YOLOv8m TRAINING CONFIGURATION FOR SEIZURE DETECTION
# Optimized for P100 GPU + Medical Surveillance
# ============================================================================
from ultralytics import YOLO
from datetime import datetime
import torch
import gc

# 🧹 Clear GPU memory
torch.cuda.empty_cache()
gc.collect()

print("="*70)
print("🔍 GPU MEMORY CHECK")
print("="*70)
if torch.cuda.is_available():
    gpu_name = torch.cuda.get_device_name(0)
    total = torch.cuda.get_device_properties(0).total_memory / 1e9
    reserved = torch.cuda.memory_reserved(0) / 1e9
    free = total - reserved
    print(f"GPU: {gpu_name}")
    print(f"Memory: {free:.2f} GB free out of {total:.2f} GB")
    if free < 10:
        print("⚠️  WARNING: YOLOv8m needs 10-12GB. May need to reduce batch!")
else:
    print("❌ No GPU detected!")
print("="*70)

# ============================================================================
# 🎯 IMPROVED CONFIGURATION - Based on Test Results (52.75% → 85%+ Target)
# ============================================================================
MODEL = "yolov8m.pt"      # 25.9M params - better for subtle seizure movements
EPOCHS = 150              # INCREASED from 100 (seizures need more training!)
IMAGE_SIZE = 640
BATCH_SIZE = 20           # P100 optimized
SAVE_PERIOD = 5           # Save every 5 epochs (Kaggle safety)
RESUME_FROM = None        # Set to checkpoint path if resuming (e.g., 'last.pt')

# Project naming
RUN_NAME = f"yolov8m_seizure_v2_{datetime.now().strftime('%Y%m%d_%H%M%S')}"

print("\n" + "="*70)
print("🚀 IMPROVED SEIZURE DETECTION TRAINING")
print("="*70)
print(f"Model:       {MODEL}")
print(f"Epochs:      {EPOCHS} ⬆️  INCREASED from 100")
print(f"Image Size:  {IMAGE_SIZE}x{IMAGE_SIZE}")
print(f"Batch Size:  {BATCH_SIZE}")
print(f"Save Period: {SAVE_PERIOD}")
print(f"Run Name:    {RUN_NAME}")
print("="*70)
print("\n📊 Key Improvements (to fix 52.75% accuracy):")
print("  • More epochs: 100 → 150")
print("  • Classification loss: 3.0 → 4.5 (main fix for 7.56% recall!)")
print("  • Box loss: 7.5 → 8.0")
print("  • Learning rate: 0.001 → 0.0008 (finer adjustments)")
print("  • Dropout: 0.22 → 0.25 (better generalization)")
print("  • Stronger augmentation (all aug params increased)")
print("  • Target: 80-85% accuracy, 70-80% seizure recall")
print("="*70)

# Load model
if RESUME_FROM:
    print(f"\n🔄 Resuming from: {RESUME_FROM}")
    model = YOLO(RESUME_FROM)
    resume = True
else:
    print(f"\n🆕 Loading: {MODEL}")
    model = YOLO(MODEL)
    resume = False

print("\n🚀 Starting training...\n")

# ============================================================================
# 🎯 TRAINING EXECUTION - SEIZURE-OPTIMIZED
# ============================================================================
results = model.train(
    # Dataset & System
    data='/kaggle/working/dataset.yaml',
    epochs=EPOCHS,
    imgsz=IMAGE_SIZE,
    batch=BATCH_SIZE,
    device=0,
    
    # Project organization
    project="seizure_detection_yolov8m",
    name=RUN_NAME,
    exist_ok=True,
    resume=resume,
    
    # Memory & Performance
    amp=True,
    workers=4,
    cache=False,
    rect=False,
    
    # Checkpoints & Logging
    save=True,
    save_period=SAVE_PERIOD,
    plots=True,
    verbose=True,
    
    # Optimizer & Learning Rate - IMPROVED
    patience=50,
    optimizer='AdamW',
    lr0=0.0008,           # REDUCED from 0.001 (slower learning for subtle features)
    lrf=0.00001,
    momentum=0.937,
    weight_decay=0.0008,  # INCREASED from 0.0005 (more regularization)
    warmup_epochs=5.0,
    warmup_momentum=0.8,
    warmup_bias_lr=0.1,
    
    # Loss Weights - OPTIMIZED (Main Fix for 7.56% recall!)
    box=8.0,              # INCREASED from 7.5 (precise localization critical)
    cls=4.5,              # INCREASED from 3.0 (CRITICAL: boost classification!)
    dfl=1.8,              # INCREASED from 1.5 (better bbox distribution)
    
    # Color Augmentation - ENHANCED (handle 24/7 lighting)
    hsv_h=0.015,          # INCREASED from 0.010 (more hue variation)
    hsv_s=0.6,            # INCREASED from 0.5 (better saturation robustness)
    hsv_v=0.5,            # INCREASED from 0.45 (day/night variations)
    
    # Geometric Augmentation - STRENGTHENED
    degrees=15.0,         # INCREASED from 10.0 (more rotation for varied positions)
    translate=0.15,       # INCREASED from 0.10 (patients move in bed)
    scale=0.6,            # INCREASED from 0.5 (more distance variation)
    shear=3.0,            # INCREASED from 2.0 (body deformation during seizures)
    perspective=0.0002,   # INCREASED from 0.00012 (camera angle variation)
    flipud=0.0,           # Never flip vertically
    fliplr=0.5,           # Horizontal flip OK
    
    # Advanced Augmentation - STRENGTHENED
    mosaic=0.5,           # INCREASED from 0.4 (more diverse contexts)
    mixup=0.15,           # INCREASED from 0.10 (better generalization)
    copy_paste=0.12,      # INCREASED from 0.10 (augment rare seizure types)
    auto_augment='randaugment',
    erasing=0.25,         # INCREASED from 0.20 (handle occlusions)
    crop_fraction=1.0,    # No cropping (need full body for seizure detection)
    close_mosaic=40,      # INCREASED from 30 (disable mosaic later for stability)
    
    # Multi-scale & Regularization - ENHANCED
    multi_scale=True,     # Different camera distances
    dropout=0.25,         # INCREASED from 0.22 (combat overfitting to training set)
    
    # Validation
    val=True,
    split='val',
    conf=0.001,
    iou=0.6,
    
    # Advanced Settings - OPTIMIZED
    seed=42,
    deterministic=False,
    single_cls=False,
    label_smoothing=0.15, # INCREASED from 0.12 (harder negative mining)
    nms=True,
    agnostic_nms=False,
    max_det=100,
)

# ============================================================================
# 📊 TRAINING COMPLETE
# ============================================================================
print("\n" + "="*70)
print("✅ TRAINING COMPLETE!")
print("="*70)
print(f"\n💾 Model Saved To:")
print(f"  seizure_detection_yolov8m/{RUN_NAME}/weights/")
print(f"    • best.pt - Best validation model")
print(f"    • last.pt - Latest checkpoint")
print(f"\n📊 Next Steps:")
print(f"  1. Validate on test set")
print(f"  2. Analyze confusion matrix")
print(f"  3. Test on real footage")
print(f"  4. Export for deployment")
print("="*70)

# Cleanup
torch.cuda.empty_cache()
gc.collect()
````

````python
# ============================================================================
# CELL 5: Quick Checkpoint Save (Run Anytime During Training)
# ============================================================================
# 🚨 RUN THIS CELL ANYTIME TO SAVE PROGRESS FOR CRASH RECOVERY
# Copies latest checkpoint to easy download location

from pathlib import Path
import shutil
from datetime import datetime


def save_checkpoint_for_download():
    """Save current checkpoint for crash recovery"""
    print("\n" + "=" * 70)
    print("💾 QUICK CHECKPOINT SAVE")
    print("=" * 70)

    latest_run = get_latest_run()
    if latest_run is None:
        print("\n❌ No training runs found yet. Start Cell 4 first.")
        return

    last_pt = latest_run / "weights/last.pt"
    if not last_pt.exists():
        print("\n❌ No checkpoint found. Training may not have produced weights yet.")
        return

    timestamp = datetime.now().strftime('%Y%m%d_%H%M%S')
    checkpoint_name = f"checkpoint_{timestamp}.pt"
    dest = Path(f"/kaggle/working/{checkpoint_name}")
    shutil.copy2(last_pt, dest)

    size = last_pt.stat().st_size / (1024 * 1024)
    print(f"\n✅ Checkpoint saved!")
    print(f"\n📦 File: {checkpoint_name}")
    print(f"💾 Size: {size:.1f} MB")
    print(f"📍 Location: {dest}")
    print("\n💡 TO RESUME TRAINING:")
    print("   1. Download this file")
    print("   2. Upload to new Kaggle session")
    print(f"   3. Set RESUME_FROM = '{checkpoint_name}' in Cell 4")
    print("   4. Run Cell 4 again")

    results_csv = latest_run / "results.csv"
    if results_csv.exists():
        results_dest = dest.with_name(f"results_{timestamp}.csv")
        shutil.copy2(results_csv, results_dest)
        print(f"\n📊 Also saved: {results_dest.name}")

    print("\n" + "=" * 70)


# Run the save
save_checkpoint_for_download()
````

````python
# ============================================================================
# CELL 5: Manual Test Set Evaluation (Skip .val())
# ============================================================================
from ultralytics import YOLO
from pathlib import Path
from collections import defaultdict

print("\n" + "=" * 70)
print("MANUAL TEST SET EVALUATION")
print("=" * 70)



weights_path = "/kaggle/working/seizure_detection/run_20251117_130236/weights/best.pt"

print(f"Using weights: {weights_path}")
model = YOLO(str(weights_path))

test_dir = "/kaggle/input/seizure-detection-yolo/yolo_seizure/test"
test_images = sorted((test_dir / "images").glob("*.jpg"))
print(f"\nFound {len(test_images)} test images")

results_by_class = defaultdict(lambda: {'correct': 0, 'total': 0})

print("\nRunning predictions on 100-sample subset (for speed)...")
for img_path in test_images[:100]:
    label_path = test_dir / "labels" / img_path.name.replace('.jpg', '.txt')
    if not label_path.exists():
        continue

    with open(label_path) as f:
        true_class = int(f.read().split()[0])

    result = model.predict(str(img_path), verbose=False)[0]
    pred_class = int(result.boxes.cls[0].cpu()) if len(result.boxes) > 0 else -1

    results_by_class[true_class]['total'] += 1
    if pred_class == true_class:
        results_by_class[true_class]['correct'] += 1

print("\n🎯 TEST SET PERFORMANCE (Sample of 100 images):")
for class_id, class_name in enumerate(["Normal", "Seizure"]):
    stats = results_by_class[class_id]
    total_cls = stats['total']
    acc = (stats['correct'] / total_cls * 100) if total_cls else 0.0
    print(f"\n  {class_name} (Class {class_id}):")
    print(f"    Correct: {stats['correct']}/{total_cls}")
    print(f"    Accuracy: {acc:.1f}%")

total_correct = sum(s['correct'] for s in results_by_class.values())
total_evaluated = sum(s['total'] for s in results_by_class.values())
overall_acc = (total_correct / total_evaluated * 100) if total_evaluated else 0.0
print(f"\n  Overall Accuracy: {overall_acc:.1f}%")
print(f"  Total evaluated: {total_evaluated} images")
````
