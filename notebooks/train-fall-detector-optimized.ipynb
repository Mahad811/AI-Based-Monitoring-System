````python
# ============================================================================
# CELL 3.5: Robust Image Loader (CRITICAL - prevents all OpenCV errors)
# ============================================================================
import cv2
import numpy as np
from PIL import Image
import ultralytics.utils.patches as yolo_patches

print("🔧 Installing bulletproof image loader...")

_FAILED_IMAGES = set()

def robust_imread(filename, flags=cv2.IMREAD_COLOR):
    """
    Robust image loader that ALWAYS returns a valid numpy array.
    Falls back to PIL, then to dummy image if everything fails.
    """
    filepath = str(filename)
    
    # Method 1: Try OpenCV with explicit buffer
    try:
        with open(filepath, 'rb') as f:
            file_bytes = np.frombuffer(f.read(), dtype=np.uint8)
        img = cv2.imdecode(file_bytes, flags)
        if img is not None and isinstance(img, np.ndarray) and img.size > 0:
            # Ensure 3 channels
            if img.ndim == 2:
                img = cv2.cvtColor(img, cv2.COLOR_GRAY2BGR)
            elif img.ndim == 3 and img.shape[2] == 1:
                img = cv2.cvtColor(img, cv2.COLOR_GRAY2BGR)
            return img
    except Exception:
        pass
    
    # Method 2: Try PIL
    try:
        pil_img = Image.open(filepath).convert('RGB')
        img = np.array(pil_img)
        img = cv2.cvtColor(img, cv2.COLOR_RGB2BGR)  # PIL uses RGB, OpenCV uses BGR
        if img is not None and isinstance(img, np.ndarray) and img.size > 0:
            return img
    except Exception:
        pass
    
    # Method 3: Return black dummy image (training continues)
    if filepath not in _FAILED_IMAGES:
        print(f"⚠️  Using dummy image for: {filepath}")
        _FAILED_IMAGES.add(filepath)
    
    # Return valid 640x640 black image
    return np.zeros((640, 640, 3), dtype=np.uint8)

# Patch Ultralytics image readers
yolo_patches.imread = robust_imread

try:
    import ultralytics.data.base as yolo_base
    yolo_base.imread = robust_imread
except Exception:
    pass

print("✅ Robust image loader installed!")
print("   - Uses OpenCV with explicit buffer")
print("   - Falls back to PIL if OpenCV fails")
print("   - Never crashes - always returns valid array")
````

````python
# ============================================================================
# CELL 3.6: Helper utilities (latest run lookup)
# ============================================================================
from pathlib import Path

PROJECT_DIR = Path("fall_detection")

def get_latest_run(project_dir: Path = PROJECT_DIR):
    """Return the most recent training run directory (or None)."""
    if not project_dir.exists():
        return None
    runs = sorted(
        [d for d in project_dir.iterdir() if d.is_dir()],
        key=lambda d: d.stat().st_mtime,
        reverse=True,
    )
    return runs[0] if runs else None

print("✅ Helper utilities loaded. Later cells will auto-detect the latest run.")
````

````python
# ============================================================================
# CELL 3.5: CRITICAL FIX - OpenCV/NumPy Compatibility
# ============================================================================
# 🚨 RUN THIS CELL BEFORE CELL 4 TO FIX OPENCV ERROR!

import cv2
import numpy as np

print("🔧 Applying OpenCV/NumPy compatibility patch...")

# Create fixed imread function
def fixed_imread(filename, flags=cv2.IMREAD_COLOR):
    """Fixed imread for Kaggle numpy/cv2 compatibility"""
    try:
        with open(str(filename), 'rb') as f:
            file_bytes = f.read()
        # Convert bytes to numpy array explicitly (this fixes the error!)
        file_array = np.frombuffer(file_bytes, dtype=np.uint8)
        im = cv2.imdecode(file_array, flags)
        # Ensure 3 dimensions
        if im is not None and im.ndim == 2:
            im = im[..., None]
        return im
    except Exception as e:
        print(f"⚠️  Error reading {filename}: {e}")
        return None

# Patch ultralytics imread BEFORE it's used
import ultralytics.utils.patches
ultralytics.utils.patches.imread = fixed_imread

# Also patch cv2.imread as backup
original_cv2_imread = cv2.imread
cv2.imread = lambda f, flags=cv2.IMREAD_COLOR: fixed_imread(f, flags)

print("✅ OpenCV patch successfully applied!")
print("✅ You can now run Cell 4 without errors")

````

````python
# ============================================================================
# ULTIMATE YOLOv8m TRAINING CONFIGURATION FOR FALL DETECTION
# Optimized for P100 GPU + Medical Surveillance
# ============================================================================
from ultralytics import YOLO
from datetime import datetime
import torch
import gc

# 🧹 Clear GPU memory
torch.cuda.empty_cache()
gc.collect()

print("="*70)
print("🔍 GPU MEMORY CHECK")
print("="*70)
if torch.cuda.is_available():
    gpu_name = torch.cuda.get_device_name(0)
    total = torch.cuda.get_device_properties(0).total_memory / 1e9
    reserved = torch.cuda.memory_reserved(0) / 1e9
    free = total - reserved
    print(f"GPU: {gpu_name}")
    print(f"Memory: {free:.2f} GB free out of {total:.2f} GB")
    if free < 10:
        print("⚠️  WARNING: YOLOv8m needs 10-12GB. May need to reduce batch!")
else:
    print("❌ No GPU detected!")
print("="*70)

# ============================================================================
# 🎯 TRAINING CONFIGURATION
# ============================================================================
MODEL = "yolov8m.pt"      # 25.9M params
EPOCHS = 200
IMAGE_SIZE = 640
BATCH_SIZE = 20           # P100 optimized
SAVE_PERIOD = 5           # Save every 5 epochs (Kaggle safety)
RESUME_FROM = None        # Set to checkpoint path if resuming

# Project naming
RUN_NAME = f"yolov8m_fall_{datetime.now().strftime('%Y%m%d_%H%M%S')}"

print("\n" + "="*70)
print("🚀 TRAINING CONFIGURATION")
print("="*70)
print(f"Model:       {MODEL}")
print(f"Epochs:      {EPOCHS}")
print(f"Image Size:  {IMAGE_SIZE}x{IMAGE_SIZE}")
print(f"Batch Size:  {BATCH_SIZE}")
print(f"Save Period: {SAVE_PERIOD}")
print(f"Run Name:    {RUN_NAME}")
print("="*70)

# Load model
if RESUME_FROM:
    print(f"\n🔄 Resuming from: {RESUME_FROM}")
    model = YOLO(RESUME_FROM)
    resume = True
else:
    print(f"\n🆕 Loading: {MODEL}")
    model = YOLO(MODEL)
    resume = False

print("\n🚀 Starting training...\n")

# ============================================================================
# 🎯 TRAINING EXECUTION
# ============================================================================
results = model.train(
    # Dataset & System
    data='/kaggle/working/dataset.yaml',
    epochs=EPOCHS,
    imgsz=IMAGE_SIZE,
    batch=BATCH_SIZE,
    device=0,
    
    # Project organization
    project="fall_detection_yolov8m",
    name=RUN_NAME,
    exist_ok=True,
    resume=resume,
    
    # Memory & Performance
    amp=True,
    workers=4,
    cache=False,
    rect=False,
    
    # Checkpoints & Logging
    save=True,
    save_period=SAVE_PERIOD,
    plots=True,
    verbose=True,
    
    # Optimizer & Learning Rate
    patience=50,
    optimizer='AdamW',
    lr0=0.001,
    lrf=0.00001,
    momentum=0.937,
    weight_decay=0.0005,
    warmup_epochs=5.0,
    warmup_momentum=0.8,
    warmup_bias_lr=0.1,
    
    # Loss Weights
    box=7.5,
    cls=2.5,
    dfl=1.5,
    
    # Color Augmentation
    hsv_h=0.012,
    hsv_s=0.6,
    hsv_v=0.45,
    
    # Geometric Augmentation
    degrees=12.0,
    translate=0.12,
    scale=0.6,
    shear=2.5,
    perspective=0.00015,
    flipud=0.0,
    fliplr=0.5,
    
    # Advanced Augmentation
    mosaic=0.5,
    mixup=0.15,
    copy_paste=0.15,
    auto_augment='randaugment',
    erasing=0.25,
    crop_fraction=1.0,
    close_mosaic=30,
    
    # Multi-scale & Regularization
    multi_scale=True,
    dropout=0.20,
    
    # Validation
    val=True,
    split='val',
    conf=0.001,
    iou=0.6,
    
    # Advanced Settings
    seed=42,
    deterministic=False,
    single_cls=False,
    label_smoothing=0.1,
    nms=True,
    agnostic_nms=False,
    max_det=100,
)

# ============================================================================
# 📊 TRAINING COMPLETE
# ============================================================================
print("\n" + "="*70)
print("✅ TRAINING COMPLETE!")
print("="*70)
print(f"\n💾 Model Saved To:")
print(f"  fall_detection_yolov8m/{RUN_NAME}/weights/")
print(f"    • best.pt - Best validation model")
print(f"    • last.pt - Latest checkpoint")
print(f"\n📊 Next Steps:")
print(f"  1. Validate on test set")
print(f"  2. Analyze confusion matrix")
print(f"  3. Test on real footage")
print(f"  4. Export for deployment")
print("="*70)

# Cleanup
torch.cuda.empty_cache()
gc.collect()

# ============================================================================
# 🧪 TEST SET EVALUATION
# ============================================================================
print("\n" + "="*70)
print("🧪 EVALUATING ON TEST SET")
print("="*70)

try:
    best_model = YOLO(f'fall_detection_yolov8m/{RUN_NAME}/weights/best.pt')
    test_results = best_model.val(
        data='/kaggle/working/dataset.yaml',
        split='test',
        imgsz=IMAGE_SIZE,
        batch=BATCH_SIZE,
        verbose=True
    )
    print(f"\n✅ Test Results:")
    print(f"  mAP50:      {test_results.box.map50:.4f}")
    print(f"  mAP50-95:   {test_results.box.map:.4f}")
    print(f"  Precision:  {test_results.box.mp:.4f}")
    print(f"  Recall:     {test_results.box.mr:.4f}")
except Exception as e:
    print(f"\n⚠️  Testing skipped: {e}")

print("\n" + "="*70)
print("🎉 ALL DONE!")
print("="*70)
````

````python
# ============================================================================
# CELL 5: Quick Checkpoint Save (Run Anytime During Training)
# ============================================================================
# 🚨 RUN THIS CELL ANYTIME TO SAVE PROGRESS FOR CRASH RECOVERY
# Copies latest checkpoint to easy download location

from pathlib import Path
import shutil
from datetime import datetime

def save_checkpoint_for_download():
    """Save current checkpoint for crash recovery"""
    
    print("\n" + "="*70)
    print("💾 QUICK CHECKPOINT SAVE")
    print("="*70)
    
    # Find latest run
    project_dir = Path("fall_detection")
    if not project_dir.exists():
        print("\n❌ No training found yet. Start training first.")
        return
    
    # Get most recent run directory
    runs = sorted([d for d in project_dir.iterdir() if d.is_dir()], 
                  key=lambda x: x.stat().st_mtime, reverse=True)
    
    if not runs:
        print("\n❌ No training runs found.")
        return
    
    latest_run = runs[0]
    last_pt = latest_run / "weights/last.pt"
    
    if last_pt.exists():
        # Copy to working directory with timestamp
        timestamp = datetime.now().strftime('%Y%m%d_%H%M%S')
        checkpoint_name = f"checkpoint_{timestamp}.pt"
        dest = f"/kaggle/working/{checkpoint_name}"
        shutil.copy2(last_pt, dest)
        
        size = last_pt.stat().st_size / (1024 * 1024)
        
        print(f"\n✅ Checkpoint saved!")
        print(f"\n📦 File: {checkpoint_name}")
        print(f"💾 Size: {size:.1f} MB")
        print(f"📍 Location: /kaggle/working/{checkpoint_name}")
        print(f"\n💡 TO RESUME TRAINING:")
        print(f"   1. Download this file")
        print(f"   2. Upload to new Kaggle session")
        print(f"   3. Set RESUME_FROM = '{checkpoint_name}' in Cell 4")
        print(f"   4. Run Cell 4 again")
        
        # Also save results.csv
        results_csv = latest_run / "results.csv"
        if results_csv.exists():
            results_dest = f"/kaggle/working/results_{timestamp}.csv"
            shutil.copy2(results_csv, results_dest)
            print(f"\n📊 Also saved: results_{timestamp}.csv")
    else:
        print("\n❌ No checkpoint found. Training may not have started yet.")
    
    print("\n" + "="*70)

# Run the save
save_checkpoint_for_download()
````

````python
# ============================================================================
# CELL 5: Manual Test Set Evaluation (Skip .val())
# ============================================================================
from ultralytics import YOLO
import cv2
import numpy as np
from pathlib import Path
from collections import defaultdict

print("\n" + "="*70)
print("MANUAL TEST SET EVALUATION")
print("="*70)

# Load model
model = YOLO("fall_detection/exp/weights/best.pt")

# Get test images
test_dir = Path("/kaggle/working/yolo_merged/test")
test_images = list(test_dir.glob("images/*.jpg"))

print(f"\nFound {len(test_images)} test images")

# Run predictions
results_by_class = defaultdict(lambda: {'correct': 0, 'total': 0})

print("\nRunning predictions...")
for img_path in test_images[:100]:  # Sample 100 for speed
    # Get ground truth from label file
    label_path = test_dir / "labels" / img_path.name.replace('.jpg', '.txt')
    if label_path.exists():
        with open(label_path) as f:
            true_class = int(f.read().split()[0])
    else:
        continue
    
    # Predict
    result = model.predict(str(img_path), verbose=False)[0]
    
    # Get predicted class
    if len(result.boxes) > 0:
        pred_class = int(result.boxes.cls[0].cpu())
    else:
        pred_class = -1  # No detection
    
    # Track accuracy
    results_by_class[true_class]['total'] += 1
    if pred_class == true_class:
        results_by_class[true_class]['correct'] += 1

# Calculate metrics
print("\n🎯 TEST SET PERFORMANCE (Sample):")
for class_id in [0, 1]:
    stats = results_by_class[class_id]
    acc = (stats['correct'] / stats['total'] * 100) if stats['total'] > 0 else 0
    class_name = "Normal" if class_id == 0 else "Seizure"
    print(f"\n  {class_name} (Class {class_id}):")
    print(f"    Correct: {stats['correct']}/{stats['total']}")
    print(f"    Accuracy: {acc:.1f}%")

total_correct = sum(s['correct'] for s in results_by_class.values())
total = sum(s['total'] for s in results_by_class.values())
overall_acc = (total_correct / total * 100) if total > 0 else 0
print(f"\n  Overall Accuracy: {overall_acc:.1f}%")
print(f"  Total evaluated: {total} images")
````
